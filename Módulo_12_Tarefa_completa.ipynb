```python
from bs4 import BeautifulSoup

pagina = BeautifulSoup(open('fortune.html', mode= 'r'), 'html.parser')
```

```python
# extraindo as linhas da tabela

tabela = pagina.find('div', {'class': 'rt-table'})
linhas = tabela.find('div', {'class': 'rt-tbody'})
```

```python
# extrair o conteudo das linhas da tabela

lista_dados = []

for linha in linhas:
    colunas = linha.find('div', {'role': 'row'})
    colunas = colunas.get_text(separator=';').split(';')
    lista_dados.append(colunas)
print(lista_dados)
```

```python
header = [ 'rank',
           'name',
           'revenues',
           'revenues-percent-change',
           'profits',
           'profits-percent-change',
           'assets',
           'mark-value',
           'employees'
]
```

```python
# escrever todas as informações extraídas para o arquivo fortune.csv

import csv

with open(file='./fortune.csv', mode='w', encoding='utf-8') as fp:
  escritor_csv = csv.writer(fp, delimiter= ',')
  escritor_csv.writerows([header] + lista_dados)
```

```python
# Criando o DataFrame

import pandas as pd

fortune_df = pd.read_csv('fortune.csv', sep=',')
```

```python
# Limpando o DataFrame:

fortune_df['revenues'] = fortune_df['revenues'].apply(lambda x: x.replace('$','') if "$" in x else x)
fortune_df['revenues-percent-change'] = fortune_df['revenues-percent-change'].apply(lambda x: x.replace('%','') if "%" in x else x)
fortune_df['profits'] = fortune_df['profits'].apply(lambda x: x.replace('$','') if "$" in x else x)
fortune_df['profits-percent-change'] = fortune_df['profits-percent-change'].apply(lambda x: x.replace('%','') if "%" in x else x)
fortune_df['assets'] = fortune_df['assets'].apply(lambda x: x.replace('$','') if "$" in x else x)
fortune_df['mark-value'] = fortune_df['mark-value'].apply(lambda x: x.replace('$','') if "$" in x else x)
```
