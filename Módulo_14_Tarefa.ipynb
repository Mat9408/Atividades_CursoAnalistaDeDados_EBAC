###1- Preço do diamante por tipo de corte

```python
import seaborn as sns
```

```python
data= sns.load_dataset('diamonds')
data.head()
```

```python
diamonds = data[['cut', 'price']].groupby('cut').agg('sum').reset_index()
grafico = sns.barplot(data = diamonds, x= 'cut', y= 'price', ci= None, palette= 'pastel')
grafico.set(title= 'Valor do diamante por tipo de corte', xlabel= 'Tipo de corte', ylabel= 'Preço do diamante')
```

Com o gráfico extraído, podemos perceber que o tipo de corte mais valioso é o 'Premium' e o que possui menor valor agregado é o 'Ideal' por ser mais caro e ser de uma raridade maior.

###2- Número de passageiros em dezembro por ano

```python
data= sns.load_dataset('flights').reset_index()
flights = data[['year', 'month', 'passengers']].query('month == "Dec"').groupby('year').agg('sum').reset_index()
flights.head()
```

```python
with sns.axes_style('whitegrid'):
  grafico= sns.lineplot(data = flights, x= 'year', y= 'passengers', palette= 'pastel')
  grafico.set(title= 'Nº de passageiro no mês de dezembro por ano', xlabel= 'Anos', ylabel= 'Nº de passageiros')
```

A cada ano subsequente, o mês de Dezembro mostra aumento na busca de passagens aéreas. Um possível motivo para o aumento no período seja pelo período de férias e festas de final de ano.
