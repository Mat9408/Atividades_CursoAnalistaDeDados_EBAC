1- Preço dos Outliers de diamante

```python
import seaborn as sns
```

```python
data= sns.load_dataset('diamonds')
data.head()
```

```python
with sns.axes_style('whitegrid'):
  grafico= sns.boxplot(x= data['price'])
  grafico.set(title= 'Preços fora da curva', xlabel= 'Preço (USD)');
```

Para situações fora da curva, o gráfico recomendado a ser utilizado é o BoxPlot. Como no exemplo acima podemos ver os preços mais exorbitantes dos diamantes giram entre 2500 a 5000 dólares.

2- Correlação entre o preço do diamante com seu peso (carat) agrupados por sua transparência (clarity)

```python
data= sns.load_dataset('diamonds')
data.head()
```

```python
diamonds= data[['price', 'carat', 'clarity']]
diamonds.head()
```

```python
with sns.axes_style('whitegrid'):
  grafico= sns.scatterplot(data= diamonds, x= 'price', y= 'carat', hue= 'clarity', palette= 'pastel');
  grafico.set(title= 'Preço do diamante pelo peso', xlabel= 'Preço (USD)', ylabel= 'Peso (g)');
  grafico.get_legend().set_title('Transparência')
  grafico.figure.set_size_inches(w= 25/2.54, h= 25/2.54)
```

De acordo com a imagem gerada, podemos concluir que os diamantes de transparência (IF- azul claro) possuem um valor maior no mercado, pois com peso de aproximadamente 1g possuem preços elevados. Já os diamantes de transparência (I1- Cinza claro) têm uma quantidae maior de gramas, mas seu valor é mais intermediário perante ao mercado.

3- Distribuição contínua do peso (weight) de carros

```python
data= sns.load_dataset('mpg')
data.head()
```

```python
carros= data[['weight']]

with sns.axes_style('whitegrid'):
  grafico= sns.histplot(data= carros, x= 'weight')
  grafico.set(title= 'Distribuição contínua aproximada do peso', xlabel= 'Peso (lb)')
```

Segundo o gráfico extraído podemos ver que na tabela possuem poucos veículos com mais de 5000 (Pounds?). Em contrapartida, há bem mais veículos com aproximadamente 2000 (pounds).
