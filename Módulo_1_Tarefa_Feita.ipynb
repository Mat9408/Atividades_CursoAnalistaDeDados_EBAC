```python
cancao = "Roda mundo, roda gigante, roda moinho, roda pião."
# primeira_maiuscula: string.title()
print(cancao.title())

# posicao: string.find(substring)
posicao = cancao.find("moinho")
print(posicao)

e_alfabeto = cancao.isalpha()
print(e_alfabeto)


```
