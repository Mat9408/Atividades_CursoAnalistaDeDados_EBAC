```python
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
penguins_df = sns.load_dataset('penguins')
```

```python
penguins_df.head(344)
```

```python
penguins_df.info()
```

###Valores nulos

```python
'''Remover todos os valores nulos do dataframe'''
penguins_df= penguins_df.dropna()
penguins_df.head()
```

###Variáveis numéricas

```python
'''Calculando a média e desvio padrão
da coluna bill_length_mm (comprimento do bico)'''

media_bill_length = penguins_df['bill_length_mm'].mean()
print(media_bill_length)

desvio_bill_length = penguins_df['bill_length_mm'].std()
print(desvio_bill_length)
```

```python
'''Criando a coluna bill_length_mm_std'''

penguins_df['bill_length_mm_std'] = penguins_df['bill_length_mm'].apply(lambda bill_length: (bill_length - media_bill_length) / desvio_bill_length)
```

```python
penguins_df.head()
```

```python
'''Calculando a média e o desvio padrão
da coluna bill_depth_mm (profundidade do bico)'''

media_bill_depth = penguins_df['bill_depth_mm'].mean()
print(media_bill_depth)

desvio_bill_depth = penguins_df['bill_depth_mm'].std()
print(desvio_bill_depth)
```

```python
'''Criando a coluna bill_depth_mm_std'''

penguins_df['bill_depth_mm_std'] = penguins_df['bill_depth_mm'].apply(lambda bill_depth: (bill_depth - media_bill_depth) / desvio_bill_depth)
```

```python
penguins_df.head()
```

```python
'''Calculando a média e desvio padrão
da coluna flipper_length_mm (tamanho da nadadeira)'''

media_flipper_length = penguins_df['flipper_length_mm'].mean()
print(media_flipper_length)

desvio_flipper_length = penguins_df['flipper_length_mm'].std()
print(desvio_flipper_length)
```

```python
'''Criando a coluna flipper_length_mm_std'''

penguins_df['flipper_length_mm_std'] = penguins_df['flipper_length_mm'].apply(lambda flipper_length: (flipper_length - media_flipper_length) / desvio_flipper_length)
```

```python
penguins_df.head()
```

```python
'''Calculando a média e desvio padrão
da coluna body_mass_g (massa corporal)'''

media_body_mass = penguins_df['body_mass_g'].mean()
print(media_body_mass)

desvio_body_mass = penguins_df['body_mass_g'].std()
print(desvio_body_mass)
```

```python
'''Criando a coluna body_mass_g_std'''

penguins_df['body_mass_g_std'] = penguins_df['body_mass_g'].apply(lambda body_mass: (body_mass - media_body_mass) / desvio_body_mass)
```

```python
penguins_df.head()
```

###Variáveis categóricas

```python
'''Transformando a coluna sex em valores numéricos inteiros
através da técnica one hot encoding'''

penguins_df['sex_m_nom'] = penguins_df['sex'].apply(lambda sex: 1 if sex == 'Male' else 0)
penguins_df['sex_f_nom'] = penguins_df['sex'].apply(lambda sex: 1 if sex == 'Female' else 0)
```

```python
penguins_df.head()
```

```python
penguins_df.head()
```

###Limpeza

```python
'''Limpando as colunas originais do DataFrame'''

penguins_df = penguins_df.drop(['island','bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g', 'sex'], axis= 1)
```

```python
penguins_df = penguins_df[['species', 'bill_length_mm_std', 'bill_depth_mm_std', 'flipper_length_mm_std', 'body_mass_g_std', 'sex_m_nom', 'sex_f_nom']]
```
