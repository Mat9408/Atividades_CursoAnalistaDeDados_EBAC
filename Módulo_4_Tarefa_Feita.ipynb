1- Funções para arquivo .csv

```python
def extrai_coluna_csv(nome_arquivo: str, indice_coluna: int, tipo_dado: str):

  limite_pessoas= []

  with open(file='./carros.csv', mode='r', encoding='utf8') as nome_arquivo:
    linha = nome_arquivo.readline()
    linha = nome_arquivo.readline()
    while linha:
      linha_separada = linha.split(sep=',')
      pessoas = linha_separada[indice_coluna]
      linha = nome_arquivo.readline()

      if tipo_dado == int:
        pessoas = int(pessoas)

        limite_pessoas.append(pessoas)
        linha = nome_arquivo.readline()

    return limite_pessoas
```

```python
limite_pessoas = extrai_coluna_csv(nome_arquivo='./carros.csv', indice_coluna= 4, tipo_dado= int)
print(limite_pessoas)
```

2- Funções para arquivo .txt

```python
def extrai_linha_txt(nome_arquivo: str, numero_linha: int):

  palavras_linha= []

  with open(file='./musica.txt', mode='r', encoding= 'utf8') as fp:
    linha = fp.readline()
    linha = fp.readline()
    while linha:
      linha_separada = linha.split(sep=' ')
      palavras_linha.append(linha_separada)
      linha = fp.readline()

  palavras_linha = palavras_linha[numero_linha]
  palavras_linha[-1] = palavras_linha[-1].strip()


  return palavras_linha
```
